# Dataset Visualization and Slicing

```python
import cfgrib
import xarray
import hvplot.xarray
import holoviews as hv
hv.extension("bokeh")
```

### Loading and Visualization South Asia ERA5 Dataset for T2M (2025 hourly)

```python
south_asia_t2m_ds = cfgrib.open_dataset('./datasets/south-asia-2025-t2m.grib', backend_kwargs={"indexpath": "./datasets/south-asia-2025-t2m.grib.idx"})
```

```python
south_asia_t2m_ds
```

```python
indices = "\n".join(south_asia_t2m_ds.dims)
columns = "\n".join(south_asia_t2m_ds.data_vars)
df_size = south_asia_t2m_ds['t2m'].size

print(f"INDICES:\n{indices}\n")
print(f"COLUMNS:\n{columns}\n")
print(f"DATAPOINTS SIZE: {df_size}")
print(f"DIMENSIONS: ")
for dim in south_asia_t2m_ds.dims:
    print(f"{dim}: {south_asia_t2m_ds.sizes[dim]}")
```

### Dropping Unnecassary Columns

```python
south_asia_t2m_modified = south_asia_t2m_ds.drop_vars(['surface','step','valid_time', 'number'])
```

```python
south_asia_t2m_modified
```

### Resampling by take 6-hour mean

```python
south_asia_t2m_modified = south_asia_t2m_modified.resample(time='6h').mean()
```

```python
indices = "\n".join(south_asia_t2m_modified.dims)
columns = "\n".join(south_asia_t2m_modified.data_vars)
df_size = south_asia_t2m_modified['t2m'].size

print(f"INDICES:\n{indices}\n")
print(f"COLUMNS:\n{columns}\n")
print(f"DATAPOINTS SIZE: {df_size}")
print(f"DIMENSIONS: ")
for dim in south_asia_t2m_modified.dims:
    print(f"{dim}: {south_asia_t2m_modified.sizes[dim]}")
```

## Slicing out Punjab Dataset

```python
PUNJAB_MIN_LAT = 27.5
PUNJAB_MAX_LAT = 34.0
PUNJAB_MIN_LONG = 69.0
PUNJAB_MAX_LONG = 75.5

punjab_t2m_modified = south_asia_t2m_modified.sel(latitude=slice(PUNJAB_MAX_LAT, PUNJAB_MIN_LAT), longitude=slice(PUNJAB_MIN_LONG, PUNJAB_MAX_LONG))
```

```python
punjab_t2m_modified
```

```python
indices = "\n".join(punjab_t2m_modified.dims)
columns = "\n".join(punjab_t2m_modified.data_vars)
df_size = punjab_t2m_modified['t2m'].size

print(f"INDICES:\n{indices}\n")
print(f"COLUMNS:\n{columns}\n")
print(f"DATAPOINTS SIZE: {df_size}")
print(f"DIMENSIONS: ")
for dim in punjab_t2m_modified.dims:
    print(f"{dim}: {punjab_t2m_modified.sizes[dim]}")
```

## Slicing out Karnatka Dataset

```python
KARNATKA_MIN_LAT = 11.5
KARNATKA_MAX_LAT = 18.5
KARNATKA_MIN_LONG = 74.0
KARNATKA_MAX_LONG = 78.5

karnatka_t2m_modified = south_asia_t2m_modified.sel(latitude=slice(KARNATKA_MAX_LAT, KARNATKA_MIN_LAT), longitude=slice(KARNATKA_MIN_LONG, KARNATKA_MAX_LONG))
```

```python
karnatka_t2m_modified
```

```python
indices = "\n".join(karnatka_t2m_modified.dims)
columns = "\n".join(karnatka_t2m_modified.data_vars)
df_size = karnatka_t2m_modified['t2m'].size

print(f"INDICES:\n{indices}\n")
print(f"COLUMNS:\n{columns}\n")
print(f"DATAPOINTS SIZE: {df_size}")
print(f"DIMENSIONS: ")
for dim in karnatka_t2m_modified.dims:
    print(f"{dim}: {karnatka_t2m_modified.sizes[dim]}")
```

## Saving Processed Dataset

```python
south_asia_t2m_modified.to_netcdf('./datasets/south_asia_t2m.nc')
punjab_t2m_modified.to_netcdf('./datasets/punjab_t2m.nc')
karnatka_t2m_modified.to_netcdf('./datasets/karnatka_t2m.nc')
```

## Visualization

```python
def visualize_data(ds:xarray.Dataset):
    da = ds['t2m']
    da_celcius = da - 273.15
    plot = da_celcius.hvplot.image(x='longitude', y='latitude', groupby='time', cmap='coolwarm')
    return plot
```

```python
display(visualize_data(south_asia_t2m_modified))
```

```python
display(visualize_data(punjab_t2m_modified))
```

```python
display(visualize_data(karnatka_t2m_modified))
```
